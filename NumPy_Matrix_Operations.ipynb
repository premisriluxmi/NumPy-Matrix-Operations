# NumPy Matrix Operations

## Objective

To create two NumPy matrices and perform addition, subtraction, matrix multiplication, and transpose operations.

```python
import numpy as np

matrix1 = np.array([
    [1, 2],
    [3, 4]
])

matrix2 = np.array([
    [5, 6],
    [7, 8]
])

print("Matrix 1:")
print(matrix1)

print("\nMatrix 2:")
print(matrix2)
```

```python
print("Shape of Matrix 1:", matrix1.shape)
print("Shape of Matrix 2:", matrix2.shape)
```

```python
addition = matrix1 + matrix2

print("Matrix Addition:")
print(addition)
```

```python
subtraction = matrix1 - matrix2

print("Matrix Subtraction:")
print(subtraction)
```

```python
multiplication = matrix1 @ matrix2

print("Matrix Multiplication:")
print(multiplication)
```

```python
transpose1 = matrix1.T
transpose2 = matrix2.T

print("Transpose of Matrix 1:")
print(transpose1)

print("\nTranspose of Matrix 2:")
print(transpose2)
```

```python
print("NumPy Matrix Operations")
print("-----------------------")

print("\nMatrix 1:")
print(matrix1)

print("\nMatrix 2:")
print(matrix2)

print("\nAddition:")
print(addition)

print("\nSubtraction:")
print(subtraction)

print("\nMatrix Multiplication:")
print(multiplication)

print("\nTranspose of Matrix 1:")
print(transpose1)

print("\nTranspose of Matrix 2:")
print(transpose2)
```

## Explanation of Results

Two matrices were created using NumPy's `np.array()` function.

Matrix addition and subtraction were performed on corresponding elements of the matrices.

Matrix multiplication was performed using the `@` operator, which multiplies rows of the first matrix by columns of the second matrix.

The transpose operation was performed using `.T`, which converts rows into columns.

These operations demonstrate basic matrix calculations commonly used in numerical computing and machine learning.
